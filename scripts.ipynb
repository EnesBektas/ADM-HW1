### Alphabet Rangoli

```python
def print_rangoli(size):
    # your code goes here
    alphabet = list("abcdefghijklmnopqrstuvwxyz")

    lines = 2 * size - 1
    line_length = 2 * ( 2 * size - 1) - 1

    for i in range(lines):
        d = size - 1 - abs(i - (size - 1))

        hyphen_length = (line_length - (1 + 4 * d)) // 2

        chars = alphabet[size - 1 - d:size]
        reverse_chars = alphabet[size - d: size]
        reverse_chars.reverse()
        chars_string = '-'.join(reverse_chars + chars)

        print(hyphen_length * "-" + chars_string + hyphen_length * "-")


if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

### The Minion Game

```python
def minion_game(string):
    # your code goes here
    kevin_score = 0
    stuart_score = 0

    for i in range(len(string)):
        if isVowel(string[i]):
            kevin_score += len(string) - i
        else:
            stuart_score += len(string) - i

    if kevin_score > stuart_score:
        print(f"Kevin {kevin_score}")
    elif stuart_score > kevin_score:
        print(f"Stuart {stuart_score}")
    else:
        print("Draw")


def isVowel(char):
    return char == 'A' or char == 'E' or char == 'I' or char == 'O' or char == 'U'

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

### Piling Up!

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
from collections import deque

N = int(input())

for i in range(N):
    L = int(input())
    condition = True
    dq = deque(map(int, input().split()))

    last_block_size = 2 ** 31

    for k in range(L):
        left_end = dq[0]
        right_end = dq[-1]
        if left_end > last_block_size or right_end > last_block_size:
            condition = False
            break
        elif left_end >= right_end:
            dq.popleft()
            last_block_size = left_end
        else:
            dq.pop()
            last_block_size = right_end

    print('Yes' if condition else 'No')
```

### Time Delta

```python
#!/bin/python3

import math
import os
import random
import re
import sys
from datetime import datetime

# Complete the time_delta function below.
def time_delta(t1, t2):
    time_format = "%a %d %b %Y %H:%M:%S %z"

    parsed_t1 = datetime.strptime(t1, time_format)
    parsed_t2 = datetime.strptime(t2, time_format)

    diff = abs(parsed_t1 - parsed_t2)
    return str(int(diff.total_seconds()))

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

```python
cube = lambda x: x ** 3# complete the lambda function

def fibonacci(n):
    # return a list of fibonacci numbers
    nums = []
    for i in range(1, n + 1):
        if i == 1:
            nums.append(0)
        elif i == 2:
            nums.append(1)
        else:
            nums.append(nums[i-3] + nums[i-2])
    return nums

if __name__ == '__main__':
    n = int(input())
    print(list(map(cube, fibonacci(n))))
```

### Hex Color Code

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
import re

regex = r'#[0-9a-fA-F]{3}(?:[0-9a-fA-F]{3})?(?![0-9a-fA-F])'
N = int(input())

begin = False
for i in range(N):
    line = input()
    if "{" in line:
        begin = True
        continue
    elif "}" in line:
        begin = False
        continue

    if begin:
        strt = 0
        while strt < len(line):
            m = re.search(regex, line[strt:])
            if m:
                print(line[m.start() + strt: m.end() + strt])
                strt += m.end()
            else:
                break
```

### Validating UID

```python
# Enter your code here. Read input from STDIN. Print output to STDOUT
import re

n = int(input())

for i in range(n):
    line = input()

    if (len(line) == 10 and
        re.match(r'^[a-zA-Z0-9]{10}$', line) and
        len(re.findall(r'[A-Z]', line)) >= 2 and
        len(re.findall(r'\d', line)) >= 3 and
        len(set(line)) == len(line)):
        print("Valid")
    else:
        print("Invalid")
```

## XML

### XML 1 - Find the Score

```python
import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
    return len(node.attrib) + sum(get_attr_number(child) for child in node)

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

### XML2 - Find the Maximum Depth

```python
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    # your code goes here
    level += 1
    if level > maxdepth:
        maxdepth = level
    for child in elem:
        depth(child, level)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

### Standardize Mobile Number Using Decorators

```python
def wrapper(f):
    def fun(l):
        # complete the function
        for i in range(len(l)):
            l[i] = "+91 " + l[i][-10: -5] + " " + l[i][-5:]
        f(l)
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l)
```

### Decorators 2 - Name Directory

```python
import operator

def person_lister(f):
    def inner(people):
        # complete the function
        sorted_people = sorted(people, key=lambda person: int(person[2]))
        return [f(person) for person in sorted_people]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```

## Number Line Jumps

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'kangaroo' function below.
#
# The function is expected to return a STRING.
# The function accepts following parameters:
#  1. INTEGER x1
#  2. INTEGER v1
#  3. INTEGER x2
#  4. INTEGER v2
#

def kangaroo(x1, v1, x2, v2):
    # Write your code here
    min_location = min(x1, x2)
    min_speed = min(v1, v2)

    if (min_speed == v1 and min_location == x1) or (min_speed == v2 and min_location == x2):
        return 'NO'

    loc_diff = abs(x1 - x2)
    speed_diff = abs(v1 - v2)

    if loc_diff % speed_diff == 0:
        return 'YES'

    return 'NO'

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    first_multiple_input = input().rstrip().split()

    x1 = int(first_multiple_input[0])

    v1 = int(first_multiple_input[1])

    x2 = int(first_multiple_input[2])

    v2 = int(first_multiple_input[3])

    result = kangaroo(x1, v1, x2, v2)

    fptr.write(result + '\n')

    fptr.close()
```

## Viral Advertising

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'viralAdvertising' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER n as parameter.
#

def viralAdvertising(n):
    # Write your code here
    shared = 5
    liked = 2
    cumulative = 2

    for i in range(n-1):
        shared = liked * 3
        liked = shared // 2
        cumulative += liked

    return cumulative

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    result = viralAdvertising(n)

    fptr.write(str(result) + '\n')

    fptr.close()
```

## Recursive Digit Sum

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'superDigit' function below.
#
# The function is expected to return an INTEGER.
# The function accepts following parameters:
#  1. STRING n
#  2. INTEGER k
#

def superDigit(n, k):
    # Write your code here
    initial_sum = sum(int(d) for d in n) * k

    return calculate(str(initial_sum))

def calculate(num):
    if len(num) == 1:
        return num
    num = str(sum([int(d) for d in num]))
    return calculate(num)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    first_multiple_input = input().rstrip().split()

    n = first_multiple_input[0]

    k = int(first_multiple_input[1])

    result = superDigit(n, k)

    fptr.write(str(result) + '\n')

    fptr.close()
```

## Insertion Sort - Part 1

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort1' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#

def insertionSort1(n, arr):
    # Write your code here
    num = arr[n-1]
    for i in range(n-2, -2, -1):
        if i == -1:
            arr[0] = num
        elif arr[i] <= num:
            arr[i+1] = num
            print(" ".join(map(str, arr)))
            break
        else:
            arr[i+1] = arr[i]
        print(" ".join(map(str, arr)))

if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort1(n, arr)
```

## Insertion Sort - Part 2
